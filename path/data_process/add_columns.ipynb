```python
import pandas as pd
import geopandas as gpd
```

```python
# 데이터 읽기
nodes_gdf = pd.read_csv("raw_data/nodes.csv")
edges_gdf = gpd.read_file("raw_data/edges.geojson")
```

```python
type_tag = ['road', 'park', 'river', 'mountain']

for col in type_tag:
    if col == 'road':
        nodes_gdf[col] = 1
        edges_gdf[col] = 1
    else:
        nodes_gdf[col] = 0
        edges_gdf[col] = 0
```

```python
nodes_gdf.to_csv("modified_data/new_nodes.csv", index=False)
edges_gdf.to_file("modified_data/new_edges.geojson", driver="GeoJSON")
```
